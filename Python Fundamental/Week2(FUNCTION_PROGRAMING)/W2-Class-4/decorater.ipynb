# Python Decorators
- Decorators are a design pattern in Python that allows a user to add new functionality to an existing object without modifying its structure.
- Decorators are very powerful and useful tool in Python since it allows programmers to modify the behavior of function or class. Decorators allow us to wrap another function in order to extend the behavior of the wrapped function, without permanently modifying it.
- In Decorators, functions are taken as the argument into another function and then called inside the wrapper function.
- Decorators can be thought of as functions which modify the functionality of another function. They help to make our code shorter and more "Pythonic".
- Decorators are usually called before the definition of a function you want to decorate.
- Decorators are used to modify the behavior of function or class. Decorators allow us to wrap another function in order to extend the behavior of the wrapped function, without permanently modifying it.


### Advantages of Decorators
- Decorators allow you to inject or modify code in functions or classes without changing the code of the function or class itself.
- Decorators allow you to reuse the same code in multiple functions or classes.
- Decorators allow you to keep your code DRY (Don't Repeat Yourself). 
- Decorators allow you to keep your code clean and readable.
- Decorators allow you to add new functionalities to the existing functions or classes.

````python
# Decorator to check if the student is a position holder (all marks above 85)
def check_position(test_position):
    def wrapper(marks):
        flag = True  # We assume all marks are above 85
        for m in marks:
            if m < 85:
                flag = False  # If any mark is below 85, they are not a position holder
                break         # No need to check further once we know they are not a position holder
        
        if flag:
            print("Congratulations, you are a position holder!")
            test_position(marks)  # Calls the main function
        else:
            test_position(marks)  # If not a position holder, just call the main function as usual
    return wrapper

# Main function to check pass or fail for each subject
@check_position
def check_result(marks):
    flag = True  # Assume the student passes all subjects
    
    for m in marks:
        if m >= 33:
            print("pass")
        else:
            print("fail")
            flag = False  # If any mark is below 33, the student failed that subject
            
    if flag:
        print("Passed all subjects!")
    else:
        print("Failed in some subjects but passed in others.")

# Get user input
enter_marks = list(map(eval, input("Enter your marks (comma separated): ").split(',')))
print("You entered marks:", enter_marks)

# Run the decorated function
check_result(enter_marks)
````

### **2. Authorization Decorator (Intermediate Challenge)**

#### **Concept**: Passing arguments to decorators.

**Challenge**: Create an `authorize` decorator that checks if a user has the required permissions to execute a function. The decorator should accept the required role as an argument.

**Example**:
```python
def authorize(role):
    def decorator(func):
        def wrapper(user, *args, **kwargs):
            if user.role != role:
                print(f"Access denied for {user.name}")
                return
            return func(user, *args, **kwargs)
        return wrapper
    return decorator

class User:
    def __init__(self, name, role):
        self.name = name
        self.role = role

@authorize("admin")
def delete_user(user, user_to_delete):
    print(f"{user.name} deleted {user_to_delete.name}")

admin = User("AdminUser", "admin")
guest = User("GuestUser", "guest")
user_to_delete = User("RandomUser", "user")

delete_user(admin, user_to_delete)  # Output: AdminUser deleted RandomUser
delete_user(guest, user_to_delete)  # Output: Access denied for GuestUser
```

#### **Real-Life Application**:
Authorization decorators are crucial in web applications, where certain functions (e.g., deleting users or updating records) should only be available to authorized users.

### **4. Retry Mechanism Decorator (Intermediate Challenge)**

#### **Concept**: Handling retries and error handling.

**Challenge**: Write a decorator that automatically retries a function if it raises an exception. The decorator should accept a number of retries and the exception to catch.

**Example**:
```python
def retry_on_failure(retries, exception):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for i in range(retries):
                try:
                    return func(*args, **kwargs)
                except exception as e:
                    print(f"Retry {i+1}/{retries} failed: {e}")
            print(f"Failed after {retries} retries")
        return wrapper
    return decorator

@retry_on_failure(retries=3, exception=ValueError)
def may_fail():
    if random.choice([True, False]):
        raise ValueError("Random failure")
    print("Success!")

may_fail()
```

#### **Real-Life Application**:
Retries are commonly used when dealing with unreliable external services (e.g., making network requests to APIs). If a network request fails due to connectivity, you might want to retry it.

### **5. Caching Results (Memoization) Decorator (Advanced Challenge)**

#### **Concept**: Caching results of expensive function calls.

**Challenge**: Write a decorator that caches the results of a function based on its arguments (also known as **memoization**). If the function is called with the same arguments again, return the cached result instead of recomputing it.

**Example**:
```python
def cache(func):
    memo = {}
    def wrapper(*args):
        if args in memo:
            print("Returning cached result")
            return memo[args]
        result = func(*args)
        memo[args] = result
        return result
    return wrapper

@cache
def expensive_computation(x, y):
    time.sleep(2)  # Simulating a slow computation
    return x + y

print(expensive_computation(3, 4))  # Output: Takes 2 seconds, then prints 7
print(expensive_computation(3, 4))  # Output: Returns instantly, prints 7
```

#### **Real-Life Application**:
Caching is a vital optimization technique, especially in data-heavy applications where certain calculations (e.g., database queries or API responses) are repeated often.

### **6. Rate Limiting Decorator (Advanced Challenge)**

#### **Concept**: Throttling function execution.

**Challenge**: Implement a decorator that limits how often a function can be called. For example, the function can only be called once every 2 seconds.

**Example**:
```python
import time

def rate_limiter(interval):
    def decorator(func):
        last_call = [0]
        def wrapper(*args, **kwargs):
            if time.time() - last_call[0] >= interval:
                last_call[0] = time.time()
                return func(*args, **kwargs)
            else:
                print("Too many requests. Please wait.")
        return wrapper
    return decorator

@rate_limiter(2)
def access_resource():
    print("Resource accessed!")

access_resource()
access_resource()  # Output: Too many requests. Please wait.
time.sleep(2)
access_resource()  # Output: Resource accessed!
```

#### **Real-Life Application**:
Rate limiting is essential in APIs and services where you need to control the number of requests a user can make within a specific period (e.g., to prevent abuse).

### **8. Class-Based Decorators (Advanced Challenge)**

class CallCounter:
    def __init__(self, func):
        self.func = func
        self.count = 0

    def __call__(self, *args, **kwargs):
        self.count += 1
        print(f"{self.func.__name__} called {self.count} times")

@CallCounter

### **9. Async Function Decorator (Advanced Challenge)**

#### **Concept**: Applying decorators to asynchronous functions (`async def`).

**Challenge**: Create a decorator that measures the time taken by asynchronous functions (functions defined with `async def`).

**Example**:
```python
import time
import asyncio

def async_timer(func):
    async def wrapper(*args, **kwargs):
        start = time.time()
        result = await func(*args, **kwargs)
        end = time.time()
        print(f"{func.__name__} took {end - start} seconds")
        return result
    return wrapper

@async_timer

    await asyncio.sleep(2)

asyncio.run(slow_async_function())  # Output: slow_async_function took 2.0 seconds
```

#### **Real-Life Application**:
