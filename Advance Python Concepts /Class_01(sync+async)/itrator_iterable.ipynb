### Iterator, Iterable, and Iteration

**Iterable**: An object capable of returning its members one at a time. Examples include lists, tuples, and strings. An object is considered iterable if it implements the `__iter__` method.

**Iterator**: An object representing a stream of data. It returns the next item of the sequence when you call the `__next__()` method on it. Iterators are objects with a `__next__()` method.

**Iteration**: The process of looping through the elements of an iterable. This is typically done using a loop (e.g., `for` loop) or by manually calling the `__next__()` method on an iterator.

Example:
```python
# Iterable
my_list = [1, 2, 3]

# Getting an iterator from an iterable
my_iterator = iter(my_list)

# Iteration
print(next(my_iterator))  # Output: 1
print(next(my_iterator))  # Output: 2
print(next(my_iterator))  # Output: 3
```

Let's go through the memory calculations for both a list created using list comprehension and a `range` object, step by step, along with an explanation of overhead.

### **1. Memory Calculation for List Comprehension**

#### Code Example:
```python
import sys

l1 = [i for i in range(100000)]
print(sys.getsizeof(l1))
```

### **Components of Memory Calculation for List**:

- **List Object Overhead**:
  - Each list in Python has some overhead due to metadata. This includes:
    - Size of the list
    - Capacity of the list
    - Reference count (for garbage collection)
  - On a 64-bit system, the overhead is typically around **64 bytes**.

- **Memory for Integers**:
  - Each integer object in Python takes up a certain amount of memory. 
  - In a typical 64-bit implementation, an integer object consumes approximately **28 bytes**.

### **Detailed Calculation**:

1. **Overhead for the List**:
   - Approximately **64 bytes**.

2. **Memory for Elements**:
   - Total number of integers: **100,000**
   - Memory for integers: 
     \[
     \text{Total Memory for Integers} = \text{Number of Integers} \times \text{Memory per Integer}
     \]
     \[
     \text{Total Memory for Integers} = 100,000 \times 28 \text{ bytes} = 2,800,000 \text{ bytes}
     \]

3. **Total Memory Calculation for List**:
   \[
   \text{Total Size} = \text{Overhead} + \text{Total Memory for Integers}
   \]
   \[
   \text{Total Size} = 64 + 2,800,000 \approx 2,800,064 \text{ bytes}
   \]
   - Note: The value you observed was **800,984 bytes**. This discrepancy is likely due to differences in the actual size of integers or the implementation details on your system.

### **2. Memory Calculation for Range Object**

#### Code Example:
```python
l2 = range(100000)
print(sys.getsizeof(l2))
```

### **Components of Memory Calculation for Range**:

- **Range Object Overhead**:
  - The `range` object has a fixed amount of overhead that contains:
    - Start value
    - Stop value
    - Step value
    - Reference count
  - The total overhead is typically around **36 to 48 bytes** (let's assume **48 bytes** for this calculation).

### **Detailed Calculation**:

1. **Overhead for the Range Object**:
   - Approximately **48 bytes**.

2. **Memory for the Range Parameters**:
   - The `range` object only stores three integers (start, stop, step), each consuming about **4 to 8 bytes** depending on the architecture:
     - Start: **4 bytes**
     - Stop: **4 bytes**
     - Step: **4 bytes**

3. **Total Memory Calculation for Range**:
   \[
   \text{Total Size} = \text{Overhead} + \text{Memory for Start} + \text{Memory for Stop} + \text{Memory for Step}
   \]
   \[
   \text{Total Size} = 48 + 4 + 4 + 4 = 60 \text{ bytes}
   \]
   - However, you saw an output of **48 bytes** because the overhead is typically included in this total, and it doesn't account for individual parameter sizes.

### **Summary of Memory Calculations**:

- **List**:
  - **Memory Used**: Approximately **800,984 bytes** (or as calculated, it could go higher depending on the system).
  - **Components**: Overhead + Size of all elements.

- **Range**:
  - **Memory Used**: Approximately **48 bytes**.
  - **Components**: Overhead only (which includes start, stop, and step internally).

### **Understanding Overhead**:
- **Overhead** refers to the extra memory consumed by data structures to maintain their metadata. For a list, this includes:
  - The list's length and allocated capacity.
  - Reference counts to manage memory.
  - Additional flags and pointers used by the Python interpreter.

### **Conclusion**:
Using `range()` is more memory-efficient than using a list comprehension to generate the same sequence of numbers. The overhead and memory consumption of a list grow significantly with the number of elements, while a range only needs a small fixed amount of memory to represent potentially large sequences of integers. # Ovrer come the memory issue

To overcome the memory issue, you can use the `range` object instead of a list. The `range` object is a memory-efficient way to represent a sequence of numbers without storing them in memory. It only stores the start, stop, and step values, along with some fixed overhead, making it more memory-efficient than a list.

Here's an example of using a `range` object instead of a list to iterate over a large sequence of numbers:

```python
# Using a range object
for i in range(100000):
    print(i)
```

In this example, the `range` object is used to generate a sequence of numbers from 0 to 99999 without storing them in memory. This approach is more memory-efficient than creating a list with the same sequence of numbers.

By using the `range` object, you can iterate over large sequences of numbers without consuming excessive memory. This can be particularly useful when dealing with large datasets or when memory usage is a concern.

# Custom Rnage Function implementation


````python
class MyRange:
    def __init__(self,start,end,step=1):
        self.start = start
        self.end = end
        self.step = step
    def __iter__(self):
        return self
    def __next__(self):
        if self.start >= self.end:
            raise StopIteration
        current = self.start
        self.start += self.step
        return current
    
r = MyRange(1,10)
for i in r:
    print(i)    
````

````python
new = MyRange(10, 10)
for i in new:
    print(i)
````

# when to use iterators in python?
- 1. Memory Efficiency: Use iterators to avoid loading large data all at once.
- 2. Lazy Evaluation: Process items only as needed, saving computation.
- 3. Infinite Sequences: Ideal for endless data sources or continuous data.
- 4. Pipeline Processing: Enables chaining transformations efficiently.
- 5. Custom Control: Use custom iterators for specific traversal needs.
