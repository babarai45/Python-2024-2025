```python
# let simple python sync and async 


import asyncio
import time

def make_coffee():
    print("making coffee")
    time.sleep(5)
    print(f'coffee is ready at {time.time()}') 

def make_toast():
    print("making toast")
    time.sleep(3)
    print("toast is ready")

def make_eggs():
    print("making eggs")
    time.sleep(5)
    print("eggs are ready")

# calling the functions in a sync way
def breakfast_sync():
    make_coffee()
    make_toast()
    make_eggs()

breakfast_sync()


# it will take 13 seconds to complete the breakfast 
```

```python
# same thing in async way
async def make_coffee_async():
    print("making coffee")
    await asyncio.sleep(5)
    print(f'coffee is ready at {time.time()}')

async def make_toast_async():
    print("making toast")
    await asyncio.sleep(3)
    print("toast is ready")

async def make_eggs_async():
    print("making eggs")
    await asyncio.sleep(5)
    print("eggs are ready")

async def breakfast_async():
    await asyncio.gather(make_coffee_async(), make_toast_async(), make_eggs_async())

await breakfast_async()

# it will take 5 seconds to complete the breakfast

# so async is faster than sync

# here is  some questions?
# 1. why async is faster than sync while both are doing the same thing with same time?
# 
# ANS: because in async way, we are not waiting for the function to complete, 
# we are just calling the function and moving to the next function, so it will take less time to complete the task 
# Actually, internally, it is using the event loop to manage the tasks, so it is faster than sync way
#
# 2. what is the use of await keyword?
#
# ANS: await keyword is used to wait for the function to complete, it is used with async functions
# internally, it is used to pause the function and move to the next function, so it is used to make the function async

# 3. what is the use of asyncio.sleep()?
#
# ANS: asyncio.sleep() is used to pause the function for the given time, it is used with async functions
# it is used to make the function async, it is used to make the function to wait for the given time

# --------------------------------------------------------------------------------

# i thing it works based on the event loop, it will call the function and move to the next function,
# and event loop will manage the tasks, 
# working of event loop is like, it will call the function and move to the next function, and it will check the previous function is completed or not
# if it is completed, it will move to the next function, otherwise it will wait for the function to complete
```

```python
# 2. await keyword
"""
This module demonstrates the use of the `await` keyword in Python's asyncio library.

The `await` keyword is used to pause the execution of an async function until the awaited function completes. It is commonly used with async functions to make them asynchronous.

Syntax:
- `await function_name()`: Waits for the function to complete.
- `await asyncio.sleep(seconds)`: Pauses the function for the given time.
- `await asyncio.gather(function1(), function2())`: Calls multiple functions concurrently.
- `await asyncio.create_task(function_name())`: Creates a task for the function, similar to threading or multiprocessing.
- `await asyncio.wait([function1(), function2()])`: Waits for multiple functions to complete.
- `await asyncio.wait_for(function_name(), timeout)`: Waits for the function to complete within the given time.
- `await asyncio.wait_until(function_name(), timeout)`: Waits for the function to complete until the given time.
- `await asyncio.wait_until_complete(function_name())`: Waits for the function to complete.

Example:
- `await make_coffee_async(a, b)`: An example async function that simulates making coffee by waiting for a calculated time.

Functions:
- `make_coffee_async(a, b)`: An async function that simulates making coffee by waiting for `a + b` seconds and then prints a message.
"""
# await keyword is used to wait for the function to complete, it is used with async functions 
# it is used to make the function async 

# syntax:
# await function_name()  # it is used to wait for the function to complete
# await asyncio.sleep(5) # it is used to pause the function for the given time
# await asyncio.gather(function1(), function2()) # it is used to call the multiple functions at a time
# await asyncio.create_task(function_name()) # it is used to create a task for the function like threading or multiprocessing or concurrent.futures
# await asyncio.wait([function1(), function2()]) # it is used to wait for the multiple functions to complete
# await asyncio.wait_for(function_name(), 5) # it is used to wait for the function to complete for the given time
# await asyncio.wait_until(function_name(), 5) # it is used to wait for the function to complete until the given time
# await asyncio.wait_until_complete(function_name()) # it is used to wait for the function to complete

# example: await with function name
async def make_coffee_async(a, b):
    c = a + b
    await asyncio.sleep(c)  # simulate waiting for the function to complete
    print(f'coffee is ready after {c} seconds')

# calling the async function
await make_coffee_async(2, 3)
```

```python
import asyncio
import time

# Define an async function with a delay
async def function_with_delay():
    print("Starting function_with_delay")
    await asyncio.sleep(7)  # 3-second delay
    print("Completed function_with_delay")

# Define an async function without a delay
async def function_without_delay():
    print("Starting function_without_delay")
    print("Completed function_without_delay")  # Executes immediately

# Main async function to compare both
async def main():
    print("Comparing functions with and without delay...\n")

    # Measure time for function with delay
    start_time = time.time()
    await function_with_delay()
    time_with_delay = time.time() - start_time

    # Measure time for function without delay
    start_time = time.time()
    await function_without_delay()
    time_without_delay = time.time() - start_time

    # Print time results
    print(f"\nTime taken with delay: {time_with_delay:.2f} seconds")
    print(f"Time taken without delay: {time_without_delay:.2f} seconds")

# Run the main function in an async environment
await main()
```

```python
class ToDoList:
    def __init__(self):
        self.tasks = []

    def add_task(self, task):
        self.tasks.append(task)
        print(f"Task '{task}' added.")

    def remove_task(self, task):
        if task in self.tasks:
            self.tasks.remove(task)
            print(f"Task '{task}' removed.")
        else:
            print(f"Task '{task}' not found.")

    def view_tasks(self):
        if self.tasks:
            print("To-Do List:")
            for i, task in enumerate(self.tasks, 1):
                print(f"{i}. {task}")
        else:
            print("To-Do List is empty.")

    def clear_tasks(self):
        self.tasks.clear()
        print("All tasks have been cleared.")

    def switch_case(self, choice, task=None):
        operations = {
            "1": self.add_task,
            "2": self.remove_task,
            "3": self.view_tasks,
            "4": self.clear_tasks
        }
        func = operations.get(choice)
        if func:
            if task:
                func(task)
            else:
                func()
        else:
            print("Invalid choice.")

# Creating ToDoList instance
todo_list = ToDoList()

# Simulate user interaction
while True:
    print("\nTo-Do List Options:")
    print("1. Add Task")
    print("2. Remove Task")
    print("3. View Tasks")
    print("4. Clear Tasks")
    print("5. Exit")
    
    choice = input("Choose an option: ")
    if choice == "5":
        print("Exiting the To-Do List app.")
        break
    elif choice in ["1", "2"]:
        task = input("Enter the task: ")
        todo_list.switch_case(choice, task)
    else:
        todo_list.switch_case(choice)
```
